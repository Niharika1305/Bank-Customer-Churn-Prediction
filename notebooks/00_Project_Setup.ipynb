# 📦 Create requirements.txt

## Purpose

This section creates the **requirements.txt** file containing all Python libraries required to run this project.

This file enables anyone to recreate the project environment using:

```bash
pip install -r requirements.txt
```

````python
requirements = """pandas
numpy
matplotlib
seaborn
scikit-learn
joblib
Flask
gunicorn
shap
"""

requirements_path = os.path.join(PROJECT_PATH, "requirements.txt")

with open(requirements_path, "w") as file:
    file.write(requirements)

print("✅ requirements.txt created successfully!")
print("Saved at:", requirements_path)
````

````python
print("Contents of requirements.txt:\n")

with open(requirements_path, "r") as file:
    print(file.read())
````

# 🌐 Create Flask Application Folder Structure

## Purpose

This section creates the folder structure required for the Flask web application.

The folders organize HTML templates, CSS, JavaScript, images, and generated reports following professional Flask project standards.

````python
# Create Flask project folders

flask_folders = [
    "flask_app",
    "flask_app/templates",
    "flask_app/static",
    "flask_app/static/css",
    "flask_app/static/js",
    "flask_app/static/images",
    "flask_app/reports"
]

for folder in flask_folders:
    os.makedirs(os.path.join(PROJECT_PATH, folder), exist_ok=True)

print("✅ Flask folder structure created successfully!")
````

````python
print("Flask Folder Structure:\n")

FLASK_PATH = os.path.join(PROJECT_PATH, "flask_app")

for root, dirs, files in os.walk(FLASK_PATH):
    level = root.replace(FLASK_PATH, "").count(os.sep)
    indent = "    " * level

    print(f"{indent}{os.path.basename(root)}/")
````
